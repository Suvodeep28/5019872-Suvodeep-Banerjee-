# **Decision Trees and SVM**

```python
X = df.drop('Target', axis=1)
y = df['Target']
```

```python
#Encode the target variables for SVM

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_encoded = le.fit_transform(y)
```

```python
dict(zip(le.classes_, le.transform(le.classes_)))
```

```python
## 1 for Dropout , 2 for Enrolled , 3 for Graduate ##
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)
```

```python
# Feature scaling for SVM

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# **MODEL TRAINING USING TREES , RANDOM FOREST ,XGBOOST**

```python
# Decision trees

from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(
    criterion='gini',
    max_depth=10,
    random_state=42
)

dt.fit(X_train, y_train)
dt_preds = dt.predict(X_test)
```

```python
# Random Forest

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)
rf_preds = rf.predict(X_test)
```

```python
# Gradient Boosting Classifier


from sklearn.ensemble import GradientBoostingClassifier

gb = GradientBoostingClassifier(
    n_estimators=150,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

gb.fit(X_train, y_train)
gb_preds = gb.predict(X_test)
```

```python
# SVM

from sklearn.svm import SVC

svm = SVC(
    kernel='rbf',
    C=1.0,
    gamma='scale',
    probability=True,
    random_state=42
)

svm.fit(X_train_scaled, y_train)
svm_preds = svm.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, classification_report

def evaluate_model(name, y_true, y_pred):
    print(f"\n{name}")
    print("Accuracy:", accuracy_score(y_true, y_pred))
    print(classification_report(y_true, y_pred, target_names=le.classes_))
```

```python
#Evaluate all models


evaluate_model("Decision Tree", y_test, dt_preds)
evaluate_model("Random Forest", y_test, rf_preds)
evaluate_model("Gradient Boosting", y_test, gb_preds)
evaluate_model("SVM", y_test, svm_preds)
```

```python
# It can be seen Random forest giving the highest accuracy
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

# **Confusion matrix for all**

```python
# Decision trees

cm_dt = confusion_matrix(y_test, dt_preds)
disp_dt = ConfusionMatrixDisplay(cm_dt, display_labels=le.classes_)
disp_dt.plot(cmap='Blues')
plt.title("Decision Tree – Confusion Matrix")
plt.show()
```

```python
# Random Forest

cm_rf = confusion_matrix(y_test, rf_preds)
disp_rf = ConfusionMatrixDisplay(cm_rf, display_labels=le.classes_)
disp_rf.plot(cmap='Blues')
plt.title("Random Forest – Confusion Matrix")
plt.show()
```

```python
# XG Boost

cm_gb = confusion_matrix(y_test, gb_preds)
disp_gb = ConfusionMatrixDisplay(cm_gb, display_labels=le.classes_)
disp_gb.plot(cmap='Blues')
plt.title("Gradient Boosting – Confusion Matrix")
plt.show()
```

```python
# SVM

cm_svm = confusion_matrix(y_test, svm_preds)
disp_svm = ConfusionMatrixDisplay(cm_svm, display_labels=le.classes_)
disp_svm.plot(cmap='Blues')
plt.title("SVM – Confusion Matrix")
plt.show()
```

# **Hyperparameter Tuning**

```python
from sklearn.model_selection import GridSearchCV
param_grid = {
    'n_estimators': [200, 300],
    'max_depth': [10, 15, None],
    'min_samples_split': [2, 5]
}
```

```python
rf_grid = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42, n_jobs=-1),
    param_grid=param_grid,
    scoring='recall_macro',
    cv=5,
    verbose=1
)

rf_grid.fit(X_train, y_train)
```

```python
# Best Model

best_rf = rf_grid.best_estimator_
print("Best Parameters:", rf_grid.best_params_)
```

# **Evaluate Tuned Model**

```python
best_rf_preds = best_rf.predict(X_test)
evaluate_model("Tuned Random Forest", y_test, best_rf_preds)
```

```python
dropout_index = list(le.classes_).index("Dropout")
dropout_index
```

```python
## Dropout Recall

from sklearn.metrics import recall_score

dropout_recall = recall_score(
    y_test,
    best_rf_preds,
    labels=[dropout_index],
    average=None
)

print("Dropout Recall:", dropout_recall[0])
```
